```python
cl_list = []
for year in range(2010, 2024):
    data1 = pd.read_csv(data_format1.format(year=year), skiprows=0)
    df = data1.iloc[1:51,2:].copy()
    df.index = data1.iloc[1:51,1]
    df = df.dropna(axis=1, how='all')
    df = df.loc[:, df.ne("(X)").any()]
    cl = pd.DataFrame(index = df.columns, columns = [year], data = np.ones(df.shape[1]))
    cl_list.append(cl)
    print(f"Year: {year}, Education Data Shape: {df.shape}")
```

```python
result = cl_list[0]
for i in range(1, len(cl_list)):
    result = result.join(cl_list[i], how='outer')
```

```python
common = result.dropna(axis=0, how='any')
```
